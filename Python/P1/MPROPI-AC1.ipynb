### Exercici 5.3 — Repte: Kindergarten
El material dona: `5804→4`, `6611→2`, `0832→3`, `9135→1`, `8046→5`. Analitza i documenta la regla. Si els exemples no permeten justificar una regla única, explica per què.

````python
# Exercici 5.1
cadena = input("Introdueix una cadena de dígits: ")

parells = 0
imparells = 0

for caracter in cadena:
    numero = int(caracter)
    if numero % 2 == 0:
        parells += 1
    else:
        imparells += 1

print(f"Parells: {parells}")
print(f"Imparells: {imparells}")
# Exercici 5.2
terme = input("Introdueix el terme inicial: ")

seguent = ""
for digit in "0123456789":
    compte = terme.count(digit)
    if compte > 0:
        seguent += f"{compte}{digit}"

print(seguent)
# Exercici 5.3 (repte)
````
